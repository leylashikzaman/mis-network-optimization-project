# Cybersecurity Attack-Path Analysis
## Quantitative Analysis & Decision Making - MIS Project

This notebook walks through the full analysis of how an attacker might navigate a corporate network using Dijkstra algorithm.

## 1. Load Libraries and Data

```python
import networkx as nx
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv("../data/network_data.csv")
print("Network Data Loaded Successfully")
df
```

## 2. Build the Network Graph

```python
G = nx.DiGraph()
for _, row in df.iterrows():
    G.add_edge(row["source"], row["target"], weight=row["vulnerability_score"])
print(f"Nodes: {G.number_of_nodes()}")
print(f"Edges: {G.number_of_edges()}")
print(list(G.nodes()))
```

## 3. Run Dijkstra Shortest Path Algorithm

```python
source_node = "Public Web Server"
target_node = "Main Database"
path = nx.dijkstra_path(G, source_node, target_node, weight="weight")
path_length = nx.dijkstra_path_length(G, source_node, target_node, weight="weight")
print("Optimal Attack Path:")
for node in path:
    print(f"  {node}")
print(f"Total Vulnerability Score: {path_length}")
```

## 4. Visualize the Network

```python
import textwrap
pos = nx.spring_layout(G, seed=42)
path_edges = list(zip(path, path[1:]))
node_colors = []
for node in G.nodes():
    if node == source_node:
        node_colors.append("#e74c3c")
    elif node == target_node:
        node_colors.append("#2c3e50")
    elif node in path:
        node_colors.append("#e67e22")
    else:
        node_colors.append("#95a5a6")
plt.figure(figsize=(14, 9))
plt.title("Cybersecurity Attack-Path Analysis", fontsize=13, fontweight="bold")
nx.draw_networkx_edges(G, pos, edge_color="lightgrey", arrows=True, arrowsize=20, width=2)
nx.draw_networkx_edges(G, pos, edgelist=path_edges, edge_color="red", arrows=True, arrowsize=25, width=3)
nx.draw_networkx_nodes(G, pos, node_color=node_colors, node_size=4500)
wrapped_labels = {node: chr(10).join(textwrap.wrap(node, width=12)) for node in G.nodes()}
nx.draw_networkx_labels(G, pos, labels=wrapped_labels, font_size=8, font_color="white", font_weight="bold")
edge_labels = nx.get_edge_attributes(G, "weight")
nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels, font_size=9)
plt.axis("off")
plt.tight_layout()
plt.show()
```

## 5. Managerial Interpretation

The shortest path algorithm reveals that the path of least resistance runs through the Phishing Email Gateway and Employee Laptop.

**Key Insight:** The weakest link is human behavior, not technical infrastructure.

**Recommendations:**
- Prioritize cybersecurity awareness training
- Implement multi-factor authentication on all employee devices
- Increase monitoring on the Admin Terminal
- The Internal Firewall is the hardest path, meaning firewall investment is working
